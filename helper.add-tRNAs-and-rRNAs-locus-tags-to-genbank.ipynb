```python
from Bio import SeqIO
contigs = []
for contig in SeqIO.parse('BBH-outputs/ASM584_to_RECOLI57/modified.gb', 'genbank'):
    contigs.append(contig)
```

```python
for contig in contigs:
    for feature in contig.features:
        # if feature.qualifiers.get('locus_tag', [None])[0] in [ 'ECMDS42_2320', 'ECMDS42_2321', 'ECMDS42_2322', 'ECMDS42_2639' ]:
            # print(feature.qualifiers.get('gene_synonym', [None])[0])
        if feature.qualifiers.get('locus_tag', [None])[0] in ['ECMDS42_2950']:
            print(feature)
```

```python
new = []
for contig in contigs:
    for feature in contig.features:
        if feature.type in ['tRNA', 'rRNA', 'misc_RNA', 'ncRNA']:
            if feature.qualifiers.get('gene_synonym', [None])[0] not in \
            ['cysT', 'rybA', 'sraB', 'ryeB', 'isrB', 'ryeC', 'ryeD', 'sroF', 'sraD', 'rygA', 'rygB', 'rygC', 'rygD', 'sraG',
             'ryhA', 'ryiA']:
                # print(feature.type, feature.qualifiers.get('locus_tag', [None])[0], feature.qualifiers.get('gene_synonym', [None])[0])
                feature.qualifiers['locus_tag'] = [feature.qualifiers.get('gene_synonym', [None])[0]]
            else:
                # print(feature.type, feature.qualifiers.get('locus_tag', [None])[0], feature.qualifiers.get('gene_synonym', [None])[1])
                feature.qualifiers['locus_tag'] = [feature.qualifiers.get('gene_synonym', [None])[1]]
                # print(feature.qualifiers)
        new.append(feature)
new
```

```python
with open('BBH-outputs/ASM584_to_C321.dA.DEPmutS.evolved_strain/modified-tRNAs-rRNAs.gb', 'w') as outfile:
    SeqIO.write(contigs, outfile, format = 'genbank')
```
