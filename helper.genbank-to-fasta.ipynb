```python
from Bio import SeqIO
contigs = []
for contig in SeqIO.parse('aaf3639_Ostrov_rE.coli-57.Design.gb', 'genbank'):
    contigs.append(contig)
```

```python
values = []
for contig in contigs:
    for feature in contig.features:
        if feature.type == 'CDS':
            values.append(feature)
```

```python
values[0].__dict__
```

```python
with open('aaf3639_Ostrov_rE.coli-57.faa', 'w') as outfile:
    for value in values:
        try:
            outfile.write('>{:s} {:s} [Escherichia coli]\n'.format(value.qualifiers['protein_id'][0], value.qualifiers.get('product', [''])[0]))
            outfile.write('{:s}\n'.format(value.qualifiers['translation'][0]))
        except:
            print(value)
```
